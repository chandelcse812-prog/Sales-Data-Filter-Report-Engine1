Import NumPy and create the data

```python
import numpy as np

months = np.arange(1, 13)

stores = np.array([
    "Store 1", "Store 2", "Store 3", "Store 4", "Store 5",
    "Store 6", "Store 7", "Store 8", "Store 9", "Store 10"
])

np.random.seed(42)

sales = np.random.randint(20000, 100000, size=120)
sales = sales.reshape(10, 12)

print("Sales data:")
print(sales)

print("Shape:", sales.shape)
print("Dimensions:", sales.ndim)
print("Total values:", sales.size)
print("Data type:", sales.dtype)
```

Monthly sales report

```python
monthly_sales = np.sum(sales, axis=0)

print("MONTHLY SALES REPORT")

for i in range(12):
    print("Month", months[i], ":", monthly_sales[i])

best_month = np.argmax(monthly_sales)
worst_month = np.argmin(monthly_sales)

print("Best Month:", months[best_month])
print("Highest Monthly Sales:", monthly_sales[best_month])

print("Lowest Sales Month:", months[worst_month])
print("Lowest Monthly Sales:", monthly_sales[worst_month])
```

Quarterly sales report using slicing

```python
q1 = sales[:, 0:3]
q2 = sales[:, 3:6]
q3 = sales[:, 6:9]
q4 = sales[:, 9:12]

quarters = np.array(["Q1", "Q2", "Q3", "Q4"])

quarterly_sales = np.array([
    np.sum(q1),
    np.sum(q2),
    np.sum(q3),
    np.sum(q4)
])

print("QUARTERLY SALES REPORT")

for i in range(4):
    print(quarters[i], ":", quarterly_sales[i])
```

Store-wise sales report

```python
store_sales = np.sum(sales, axis=1)

print("STORE-WISE SALES REPORT")

for i in range(10):
    print(stores[i], ":", store_sales[i])

best_store = np.argmax(store_sales)
worst_store = np.argmin(store_sales)

print("Best Store:", stores[best_store])
print("Highest Store Sales:", store_sales[best_store])

print("Lowest Performing Store:", stores[worst_store])
print("Lowest Store Sales:", store_sales[worst_store])
```

Boolean masking and combined conditions

```python
# Months with sales above the target
target = 600000
target_mask = monthly_sales > target

print("Months above target:")
print(months[target_mask])
print("Their sales:")
print(monthly_sales[target_mask])

# Stores with below-average annual sales
average_sales = np.mean(store_sales)
below_average_mask = store_sales < average_sales

print("Average Store Sales:", average_sales)
print("Stores below average:")
print(stores[below_average_mask])

# Sales greater than 5 lakhs AND less than 7 lakhs
and_mask = (monthly_sales > 500000) & (monthly_sales < 700000)

print("Months between 5 and 7 lakhs:")
print(months[and_mask])

# Sales greater than 7 lakhs OR less than 4 lakhs
or_mask = (monthly_sales > 700000) | (monthly_sales < 400000)

print("Months above 7 lakhs or below 4 lakhs:")
print(months[or_mask])

# Stores that are not below average
not_below_average = ~below_average_mask

print("Stores at or above average:")
print(stores[not_below_average])
```

Fancy indexing and top 3 months

```python
selected_stores = sales[[1, 4, 7]]

print("Sales of selected stores:")
print(selected_stores)

top_month_indexes = np.argsort(monthly_sales)[-3:][::-1]

print("Top 3 Months:")
print(months[top_month_indexes])

print("Their Sales:")
print(monthly_sales[top_month_indexes])
```

Growth targets using np.linspace()

```python
growth_targets = np.linspace(400000, 800000, 12)

target_achieved = monthly_sales >= growth_targets

print("MONTHLY GROWTH TARGET REPORT")

for i in range(12):
    print(
        "Month:", months[i],
        "| Sales:", monthly_sales[i],
        "| Target:", round(growth_targets[i]),
        "| Achieved:", target_achieved[i]
    )
```

Final project summary

```python
total_sales = np.sum(sales)
average_monthly_sales = np.mean(monthly_sales)

print("FINAL SALES SUMMARY")
print("-------------------")
print("Total Annual Sales:", total_sales)
print("Average Monthly Sales:", round(average_monthly_sales))
print("Best Month:", months[best_month])
print("Best Store:", stores[best_store])
print("Best Quarter:", quarters[np.argmax(quarterly_sales)])
print("Best Quarter Sales:", np.max(quarterly_sales))
```
